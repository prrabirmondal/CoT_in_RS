```python

def read_json_file(filepath):
    """Reads a JSON file and returns the data as a Python object (dict or list)."""
    try:
        with open(filepath, 'r') as file:
            data = json.load(file)
        return data
    except FileNotFoundError:
        print(f"Error: File not found at {filepath}")
        return None
    except json.JSONDecodeError:
        print(f"Error: Invalid JSON format in {filepath}")
        return None
    
```

```python
# All Users

user_file = '/DATA/punitsingh1801/prabir/prabir/python_program/Earl_workshop/GoT/data/users.json'
user_file = read_json_file(user_file)


all_users = []
for u_id in list(user_file.keys()):
    dic = {"user_id": u_id, **user_file[u_id]}
    all_users.append(dic)
    
all_users = pd.DataFrame(all_users)
print(f"total users = {len(all_users)}")
```

```python

# All Movies

movie_file = '/DATA/punitsingh1801/prabir/prabir/python_program/Earl_workshop/GoT/data/movies.json'
movie_file = read_json_file(movie_file)


all_movies = []
for m_id in list(movie_file.keys()):
    dic = {"movie_id": m_id, **movie_file[m_id]}
    all_movies.append(dic)
    
all_movies = pd.DataFrame(all_movies)

print(f"total movies = {len(all_movies)}")
# all_movies.head()
```

```python
## All Movies

clusters_file = '/DATA/punitsingh1801/prabir/prabir/python_program/Earl_workshop/GoT/data/clusters.json'
clusters_file = read_json_file(clusters_file)

all_cluster = []
for user in clusters_file:
    keys = list(user.keys())
    user_id = int(user[keys[0]])
    user_clusters = user[keys[1]]

    for user_cluster in user_clusters:
       
        cluster = {"user_id": user_id, **user_cluster}
        all_cluster.append(cluster)
        
all_cluster = pd.DataFrame(all_cluster)
all_cluster = all_cluster.sort_values(by=["user_id", "clusterID"])

print(f"total user_clusters = {len(all_cluster)}")
```

```python
# All Users

movie_imdb_id_file = '/DATA/punitsingh1801/prabir/prabir/python_program/Earl_workshop/GoT/data/imdb_id_map.json'
movie_imdb_id_file = read_json_file(movie_imdb_id_file)


all_movies = []
for m_id in list(movie_imdb_id_file.keys()):
    # print(m_id)
    dic = (movie_imdb_id_file[m_id])
    all_movies.append(dic)
    
all_movies = pd.DataFrame(all_movies)
print(f"total users = {len(all_movies)}")
```

```python
# all_movies.head()
```

```python
# All reviews

review_file = '/DATA/punitsingh1801/prabir/prabir/python_program/Earl_workshop/GoT/data/movie_reviews/grouped_reviews.json/grouped_reviews.json'
review_file = read_json_file(review_file)

all_movie_reviews = []
for m_id in list(review_file.keys()):
    for review in review_file[m_id]:
        r = {"movie_id": m_id, **review}
        all_movie_reviews.append(r)

all_movie_reviews = pd.DataFrame(all_movie_reviews)
# all_movie_reviews.head()
```

```python
reviews = pd.merge(all_movies, all_movie_reviews, right_on="movie_id", left_on="imdb_id", how="left")
```

```python
reviews = reviews.iloc[:, [0, 1, 2, 3, 5, 6, 7, 9]]
reviews = reviews.rename(columns={"movie_id_x": "movie_id"}) # rename column names
reviews = reviews[["movie_id", "imdb_id", "original_title", "clean_title", "review_title", "review_rating", "review_text", "review_date"]] # reorder the columns
```
