```python
import numpy as np, pandas as pd
```

```python
ratings = pd.read_csv("/DATA/punitsingh1801/prabir/prabir/python_program/Earl_workshop/GoT/data/ratings.dat", sep="::", engine="python")
ratings.columns = ["user_id", "movie_id", "rating", "timestamp"]

print(f"total ratings = {len(ratings)}")
ratings.head()
```

```python
# movie Vs. total_users
movie_ids = list(set(ratings["movie_id"]))
user_ids = list(set(ratings["user_id"]))
# print(len(movie_ids))
users_count, u_count = [], []

# check movie_id wise user count
for m_id in movie_ids:
    # print(m_id)
    users = list(ratings[ratings["movie_id"] == m_id]["user_id"])
    # users = list(ratings[ratings["movie_id" == m_id]]["user_id"])
    total_users = len(users)
    u_count.append(total_users)
    dic = {"movie_id": m_id, "total_users": total_users, "users": users}
    users_count.append(dic)
    
    # print(m_id, total_users, users)

users_count = pd.DataFrame(users_count)

u_count = np.array(u_count)
# print(max(u_count), min(u_count))

# calculate the mean and the std of the user count per movie
mean, std = np.mean(u_count), np.std(u_count)
# print(f"mean of users per movie = {mean}. \n std of users per movie = {std}")
lower_range, upper_range = abs(mean-std), (mean + std)
# print(f"range of the first standard deviation: {lower_range} to {upper_range}")

# select only the movie having user count in the first standard deviation range
selected_movies = list(users_count[(users_count["total_users"] >= lower_range) & (users_count["total_users"] <= upper_range)]["movie_id"])
```

```python

# user Vs. total_movies
movie_ids = list(set(ratings["movie_id"]))
user_ids = list(set(ratings["user_id"]))
# print(len(movie_ids))
movies_count, m_count = [], []


# Calculate the user_id wise movies count
for u_id in user_ids:
    # print(m_id)
    movies = list(ratings[ratings["user_id"] == u_id]["movie_id"])
    # users = list(ratings[ratings["movie_id" == m_id]]["user_id"])
    total_movies = len(movies)
    m_count.append(total_movies)
    dic = {"user_id": u_id, "total_movies": total_movies, "movies": movies}
    movies_count.append(dic)
    
    # print(m_id, total_users, users)

movies_count = pd.DataFrame(movies_count)

m_count = np.array(m_count)
# print(max(u_count), min(u_count))

# calculate the mean and std of the movies count per user
mean, std = np.mean(m_count), np.std(m_count)
# print(f"mean of users per movie = {mean}. \n std of users per movie = {std}")
lower_range, upper_range = abs(mean-std), (mean + std)
# print(f"range of the first standard deviation: {lower_range} to {upper_range}")

# select the users having the movie count in the range of first standard deviation
selected_users = list(movies_count[(movies_count["total_movies"] >= lower_range) & (movies_count["total_movies"] <= upper_range)]["user_id"])
```

```python

# print(f"selected total users = {len(selected_users)}, and \nselected total movies = {len(selected_movies)}")
# filter the ratings matrix and keep only those record of movies and users present in the selected users and selected movies
# 
filtered_ratings = []
for indx, row in ratings.iterrows():
    if row["user_id"] in selected_users and row["movie_id"] in selected_movies:
        dic = {"user_id": row["user_id"], "movie_id" : row["movie_id"], "rating" : row["rating"]}
        filtered_ratings.append(dic)
        
filtered_ratings = pd.DataFrame(filtered_ratings)

# print(len(filtered_ratings), len(ratings))
# print((len(filtered_ratings)/len(ratings))*100)
# filtered_ratings.to_csv("filtered_ratings.csv", index=False)
```
